## 1.1 Contexto

Este projeto insere-se na área de Recursos Humanos e mercado de trabalho, sendo relevante para organizações, empresas de recrutamento e departamentos de RH interessados em compreender a mobilidade profissional.

O dataset contém dados históricos de natureza demográfica, académica e profissional sobre um conjunto de candidatos, bem como informação sobre a sua intenção de mudar de emprego.

Conhecer os perfis com maior intenção de procurar uma nova oportunidade pode ajudar a compreender tendências do mercado de trabalho e apoiar estratégias de recursos humanos.



## 1.2 Problema de negócio

O problema em causa consiste em perceber se através de um conjunto de características disponíveis sobre um candidato é possível prever a sua intenção de mudar de emprego.

Isto é importante para a tomada de decisões acerca de atração, retenção e desenvolvimento de talento por parte das empresas e departamentos de RH, permitindo identificar perfis com maior propensão para a procura de uma mudança profissional.

De forma geral, os objetivos deste projeto passam por compreender os dados, identificar padrões associados à intenção de mudança de emprego, preparar os dados para modelação, treinar e comparar modelos adequados ao problema e avaliar a capacidade dos modelos para generalizar para candidatos que não foram utilizados durante o treino.

O objetivo final é que, quando a resposta sobre a intenção de mudança de emprego não estiver disponível para um novo candidato, seja possível estimá-la a partir das restantes características conhecidas.

A previsão não representa uma certeza sobre a intenção da pessoa. Trata-se de uma estimativa baseada nos padrões encontrados nos dados históricos.


## 1.3 Variável-alvo

A variável-alvo do projeto é **`procura_mudanca_emprego`**.

Esta variável indica se o candidato procura ou não uma mudança de emprego.

## 1.7 Métricas e critérios de sucesso

A métrica principal é o **F1 da classe Sim (1)**, para equilibrar precision e recall.
A accuracy é complementar porque prever sempre Não pode acertar na maioria dos casos.
Usamos ainda recall, precision, ROC-AUC e **Average Precision (AP)**, que resume a curva precision-recall sem ser a sua área trapezoidal.

Os critérios abaixo foram definidos nesta revisão, antes de executar a modelação.
São metas pedagógicas de avaliação; não representam um requisito comercial validado.

| Critério | Meta | Justificação |
|---|---|---|
| F1 da classe Sim | ≥ 0,50 no teste | Equilíbrio inicial entre detetar positivos e evitar falsos alarmes |
| Recall da classe Sim | ≥ 0,60 no teste | Identificar pelo menos 60% dos candidatos que declaram procurar mudança |
| ROC-AUC | ≥ 0,75 no teste | Discriminação superior à referência aleatória |
| Superar o baseline | F1 e AP superiores ao DummyClassifier | Demonstrar utilidade para a classe minoritária |
| Interpretabilidade | Identificar e discutir as 3 variáveis mais relevantes | Apoiar a leitura dos resultados |
| Questões de negócio | Responder a Q1, Q2 e Q3 com evidência | Relacionar análise e problema inicial |

Seleciona-se o modelo pela média do F1 em validação cruzada estratificada de cinco folds, apenas no treino. O limiar fica em **0,50**, definido antes de consultar o teste. O teste confirma o desempenho final e não escolhe modelos ou limiares.

## 1.8 Recursos, riscos e restrições

O CSV fornecido tem dados de candidatos. A sua origem pública, licença, método de amostragem e data de recolha não estão documentados nos ficheiros; não se pode afirmar que a amostra representa o mercado de trabalho.

O enunciado prevê a utilização de um CSV fornecido pelo formador como alternativa à recolha própria. Neste trabalho usa-se o ficheiro disponibilizado, sem inventar uma recolha por web scraping ou API. A origem e a eventual validação pelo formador devem ser confirmadas pelo grupo.

Os principais riscos são os valores em falta, o desequilíbrio entre classes, a falta de datas e a possível mudança de contexto. O género é uma variável sensível a avaliar na análise de erros. Uma intenção declarada de mudança não equivale a uma saída efetiva da empresa. O modelo serve para exploração e apoio humano; a adequação a decisões reais exige validação adicional.

## 1.9 Impacto esperado

O projeto poderá contribuir para uma melhor compreensão dos fatores associados à intenção de mudança de emprego e apoiar a tomada de decisão na área de Recursos Humanos.

Os principais benefícios esperados são:

- *Apoio à tomada de decisão*, através da identificação de candidatos com maior probabilidade de procurar mudar de emprego;
- *Identificação de padrões relevantes* relacionados com a intenção de mudança profissional;
- *Redução de risco*, ajudando a estudar a intenção de mudança, sem prever diretamente saídas efetivas;
- *Otimização de recursos*, apoiando estratégias de retenção e gestão de talento;
- *Apoio ao desenvolvimento de estratégias de Recursos Humanos* baseadas nos padrões identificados nos dados.
