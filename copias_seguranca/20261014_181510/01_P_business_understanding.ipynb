## 1.1 Contexto

Este projeto insere-se na área de Recursos Humanos e mercado de trabalho, sendo relevante para organizações, empresas de recrutamento e departamentos de RH interessados em compreender a mobilidade profissional.

O dataset contém dados históricos de natureza demográfica, académica e profissional sobre um conjunto de candidatos, bem como informação sobre a sua intenção de mudar de emprego.

A análise destes dados permite conhecer os perfis com maior intenção de procurar uma nova oportunidade pode ajudar a compreender tendências do mercado de trabalho e apoiar estratégias de recursos humanos.



## 1.2 Problema de negócio

O problema em causa consiste em perceber se através de um conjunto de características disponíveis sobre um candidato é possível prever a sua intenção de mudar de emprego.

Isto é importante para a tomada de decisões acerca de atração, retenção e desenvolvimento de talento por parte das empresas e departamentos de RH, permitindo identificar perfis com maior propensão para a procura de uma mudança profissional.

De forma geral, os objetivos deste projeto passam por compreender os dados, identificar padrões associados à intenção de mudança de emprego, preparar os dados para modelação, treinar e comparar modelos adequados ao problema e avaliar a capacidade dos modelos para generalizar para candidatos que não foram utilizados durante o treino.

O objetivo final é que, quando a resposta sobre a intenção de mudança de emprego não estiver disponível para um novo candidato, seja possível estimá-la a partir das restantes características conhecidas.

A previsão não representa uma certeza sobre a intenção da pessoa. Trata-se de uma estimativa baseada nos padrões encontrados nos dados históricos.


## 1.3 Variável-alvo

A variável-alvo do projeto é **`procura_mudanca_emprego`**.

Esta variável indica se o candidato procura ou não uma mudança de emprego.

## 1.7 Métricas e critérios de sucesso

**Métricas**:

A escolha das métricas concretas será aprofundada depois de compreender a distribuição da variável-alvo. 

Caso exista desequilíbrio entre as classes, a **accuracy** não deverá ser analisada isoladamente, sendo também relevantes métricas como **precision, recall e F1-score**.

**Critérios**:

O projeto será considerado bem-sucedido se for possível construir um modelo que apresente capacidade de **generalização**, conseguindo distinguir de forma útil os candidatos que procuram mudar de emprego daqueles que não procuram em dados que não foram utilizados no treino.

| Critério | Meta | Justificação |
|---|---|---|
| Desempenho do modelo | ex.: F1 ≥ 0,75 · R² ≥ 0,70 · MAE < 10% da média | |
| Superar o baseline | ex.: melhor do que prever sempre a classe maioritária / a média | |
| Interpretabilidade | ex.: identificar as 3 variáveis mais influentes | |
| Resposta às questões de negócio | Q1, Q2, Q3 respondidas com evidência | |



Para além do desempenho estatístico, os resultados deverão ser interpretáveis no contexto do problema inicial e permitir retirar conclusões úteis sobre os fatores associados à intenção de mudança de emprego.




## 1.8 Recursos, riscos e restrições

- Dados disponíveis: dataset com dados individuais de candidatos numa amostra representativa cuja qualidade será avaliada em fase posterior. 

- Riscos e restrições: Os principais riscos e restrições estão relacionadas com a qualidade e disponibilidade dos dados, a existência de valores em falta, a privacidade da informação, o tempo disponível para o desenvolvimento do projeto e a possibilidade (a avaliar) de data leakage.

## 1.9 Impacto esperado

O projeto poderá contribuir para uma melhor compreensão dos fatores associados à intenção de mudança de emprego e apoiar a tomada de decisão na área de Recursos Humanos.

Os principais benefícios esperados são:

- *Apoio à tomada de decisão*, através da identificação de candidatos com maior probabilidade de procurar mudar de emprego;
- *Identificação de padrões relevantes* relacionados com a intenção de mudança profissional;
- *Redução de risco*, permitindo antecipar possíveis situações de saída;
- *Otimização de recursos*, apoiando estratégias de retenção e gestão de talento;
- *Apoio ao desenvolvimento de estratégias de Recursos Humanos* baseadas nos padrões identificados nos dados.
